```python
from pydantic_ai import Agent
from pydantic_ai_jupyter import DisplayAgent, ToolView
from html import escape
from typing import Any

# Create an agent with a code execution tool
agent = Agent("anthropic:claude-sonnet-4-20250514")

@agent.tool_plain
def execute_code(code: str) -> str:
    """Execute Python code and return the result."""
    # Fake execution for demo
    return f"Executed: {len(code)} chars of code"

# Custom view that shows the code nicely while streaming
class CodeExecutionView(ToolView):
    def render_streaming_call(
        self,
        tool_name: str,
        args: dict[str, Any] | str,  # Now receives parsed dict!
        tool_call_id: str | None = None,
    ) -> str:
        if isinstance(args, dict):
            code = args.get("code", "")
        else:
            code = str(args)
        
        return f"""
        <div style="border: 1px solid #6366f1; border-radius: 8px; overflow: hidden;">
            <div style="background: #6366f1; color: white; padding: 8px 12px; font-weight: 500;">
                ⏳ Writing code...
            </div>
            <pre style="margin: 0; padding: 12px; background: #1e1e1e; color: #d4d4d4; font-size: 13px; overflow-x: auto;">{escape(code)}<span style="animation: blink 1s infinite;">▊</span></pre>
        </div>
        <style>@keyframes blink {{ 50% {{ opacity: 0; }} }}</style>
        """

    def render_call(
        self,
        tool_name: str,
        args: dict[str, Any] | str,
        tool_call_id: str | None = None,
    ) -> str:
        code = args.get("code", "")
        
        return f"""
        <div style="border: 1px solid #22c55e; border-radius: 8px; overflow: hidden;">
            <div style="background: #22c55e; color: white; padding: 8px 12px; font-weight: 500;">
                ✓ Code ready to execute
            </div>
            <!-- <pre style="margin: 0; padding: 12px; background: #1e1e1e; color: #d4d4d4; font-size: 13px; overflow-x: auto;">{escape(code)}</pre> -->
        </div>
        """

    def render_result(
        self,
        tool_name: str,
        content: Any,
        tool_call_id: str | None = None,
        is_retry: bool = False,
        max_length: int = 500,
    ) -> str:
        return f"""
        <div style="border: 1px solid #10b981; border-radius: 8px; padding: 12px; background: #ecfdf5;">
            <div style="color: #047857; font-weight: 500;">📤 Output</div>
            <pre style="margin: 8px 0 0 0; color: #065f46;">{escape(str(content))}</pre>
        </div>
        """

# Create DisplayAgent with custom view
display_agent = DisplayAgent(agent, tool_views={"execute_code": CodeExecutionView()})
```

```python
result = await display_agent.run("Write a 3 functions to wow me")
```
